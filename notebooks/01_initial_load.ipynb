# 01 — Initial Load

Builds all static dimension tables from the source files:
- **DIM_DATE** — generated calendar (2026-01-01 → 2027-12-31)
- **DIM_TRANSPORT_TYPE** — hard-coded CO2 factors
- **DIM_EQUIPMENT** — CO2 reference CSV
- **DIM_CITY** — org-site cities from PERSONNEL files
- **DIM_STAFF** — staff from all 6 PERSONNEL files with job normalization

```python
import sys

sys.path.insert(0, "../src")

from config.settings import ETLConfig
from jobs.initial_load import run_initial_load
from utils.spark import get_spark

spark = get_spark()
config = ETLConfig()
```

```python
tables = run_initial_load(spark, config)
print("Initial load complete.")
```

## DIM_DATE

```python
print(f"DIM_DATE: {tables.dim_date.count()} rows")
tables.dim_date.show(5)
```

## DIM_TRANSPORT_TYPE

```python
print(f"DIM_TRANSPORT_TYPE: {tables.dim_transport_type.count()} rows")
tables.dim_transport_type.show()
```

## DIM_EQUIPMENT

```python
print(f"DIM_EQUIPMENT: {tables.dim_equipment.count()} rows")
tables.dim_equipment.show(10)
```

## DIM_CITY

```python
print(f"DIM_CITY: {tables.dim_city.count()} rows")
tables.dim_city.show()
```

## DIM_STAFF

```python
print(f"DIM_STAFF: {tables.dim_staff.count()} rows")
tables.dim_staff.show(10)
print("\nJob title distribution:")
tables.dim_staff.groupBy("JOB_TITLE").count().orderBy("JOB_TITLE").show()
print("\nStaff per site:")
tables.dim_staff.groupBy("SK_SITE").count().orderBy("SK_SITE").show()
```
